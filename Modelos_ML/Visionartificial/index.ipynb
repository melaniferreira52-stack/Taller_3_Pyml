### Detencción de rostros con Python usando OpenCV

```python
#1. Instalación de OpenCV Library

#pip install opencv-python

#%pip uninstall -y opencv-python opencv-python-headless opencv-contrib-python opencv-contrib-python-headless

#%pip install opencv-python==5.0.0.93

import cv2

#cargar imagen

imgpath = 'imagen_rostros.jpg'

#4. Leer la imagen
img = cv2.imread(imgpath)

#5. Mostar la imagen
img.shape

```

```python
# 6. Convertir laimagen a escala de grises
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_img.shape
```

```python

# Cargar clasificador de rostros
cascade = "haarcascade_frontalface_default.xml"
face_cascade = cv2.CascadeClassifier(cascade)

# Verificar si el clasificador cargo correctamente
if face_cascade.empty():
    print("Error: No se pudo cargar, verificar el xml")
else:
    print("Clasificador cargado correctamente")    
```

```python
import cv2
print(cv2.__file__)   # ¿de dónde se está importando?
print(cv2.__version__)
print(dir(cv2))       # ¿aparece 'CascadeClassifier' en la lista?
```

```python
# 9. Se realiza la deteccion del rostro en laimagen en escala de grises. Para ellos se hace uso de la funcion detectMultiscale(),esta se utiliza para detectar caras de diferentes tamaños de la imagen de entrada

face = face_cascade.detectMultiScale(gray_img, scaleFactor= 1.1, minNeighbors= 2, minSize=(40,40))
```

```python
# 10. For para recorrer todas las caras y dibujar un rectángulo sobre la imagen original (img)

for (x,y,w,h) in face:
    cv2.rectangle(img, (x,y), (x+y, w+h), (0, 0, 225), 2)
```

```python
# 11. Las imagenes

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
# 12. Importar la matplotlib

import matplotlib.pyplot as plt

plt.figure(figsize=(20,10))
plt.imshow(img_rgb)
plt.axis('off')
```
