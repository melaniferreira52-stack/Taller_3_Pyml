# **Cargar API Pandas** 💗

Pandas es una de las bibliotecas más populares de Python para manipular, analizar y procesar datos. 

Para qué sirve
* Leer y escribir archivos: Permite importar y exportar datos en formatos como CSV, Excel, JSON y bases de datos SQL.
* Limpieza de datos: Facilita la detección, eliminación o relleno de valores faltantes o nulos.
* Transformación y análisis: Permite filtrar, agrupar, ordenar y combinar grandes volúmenes de información de manera rápida y eficiente.

```python
#pip install pandas
#pip install numpy
#pip install requests

import requests
import pandas as pd 

#importar datos de una API
url = "https://rickandmortyapi.com/api/character"
data = requests.get(url).json()

data_api = pd.DataFrame(data['results'])

data_api = data_api[['id', 'name', 'status', 'species', 'type', 'gender']]

print(data_api.head())

```
